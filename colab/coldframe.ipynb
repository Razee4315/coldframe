# coldframe on Google Colab

Render a Remotion project on a free Colab runtime and save the MP4 to Google Drive.
Your laptop stays cold.

* Run the cells top to bottom, or let Claude Code drive this notebook through the
  [Colab MCP server](https://github.com/googlecolab/colab-mcp).
* Every call returns immediately and the work runs in the background, so poll with `cf.status()`.
* **Private repo?** Add a Colab secret called `GITHUB_TOKEN` (key icon, left sidebar) and let this notebook read it.

For the fastest final renders, use the GitHub Actions workflow instead (8 machines in parallel).
Colab is best for experiments, GPU tests and long single-machine jobs.

```python
#@title 1. Settings
REPO = "Razee4315/coldframe"      #@param {type:"string"}
REF = "main"                      #@param {type:"string"}
PROJECT_DIR = "example"           #@param {type:"string"}
COMPOSITION = "ColdframePromo"    #@param {type:"string"}
DRIVE_FOLDER = "coldframe"        #@param {type:"string"}
```

```python
#@title 2. Load coldframe and check the machine
!curl -fsSL https://raw.githubusercontent.com/Razee4315/coldframe/main/colab/coldframe_colab.py -o coldframe_colab.py
import importlib, coldframe_colab as cf
importlib.reload(cf)
cf.gpu()
```

```python
#@title 3. Install (2-4 min, background)
cf.setup(REPO, ref=REF, project_dir=PROJECT_DIR)
```

```python
#@title 4. Check progress (run again until it says "ready" / "done")
cf.status()
```

```python
#@title 5. Render (background)
cf.render(COMPOSITION)
```

```python
#@title 6. Check progress
cf.status()
```

```python
#@title 7. Save to Google Drive (asks for Drive access the first time)
cf.to_drive(DRIVE_FOLDER)
```
